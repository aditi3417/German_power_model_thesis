```python



def get_type(event_str):
    if pd.isna(event_str) or event_str == "":
        return ""
    ids     = [i.strip() for i in event_str.split(";")]
    types   = [type_map.get(i, "") for i in ids]
    types   = [t for t in types if t != ""]
    if not types:
        return ""
    planned = types.count("Planned")
    forced  = types.count("Forced")
    return "Planned" if planned > forced else "Unplanned"



print("\n Building Type column...")
outages["Type_entsoe_ex_post"] = outages[event_col].apply(get_type)
print(outages["Type_entsoe_ex_post"].value_counts())
```

```python


no_event  = outages[event_col].isna() | (outages[event_col] == "")
has_event = ~no_event
got_type  = outages["Type_entsoe_ex_post"] != ""

print(f"\nTotal rows:                    {len(outages):,}")
print(f"No event (unit available):     {no_event.sum():,}")
print(f"Has event:                     {has_event.sum():,}")
print(f"Has event + got type:          {(has_event & got_type).sum():,}  ← should be high")
print(f"Has event but no type:         {(has_event & ~got_type).sum():,}  ← should be low")

```
